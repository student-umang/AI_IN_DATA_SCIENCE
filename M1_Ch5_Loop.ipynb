Write a program to reverse a given number using a while loop

```python
#WAP to reverse a given number using a while loop
num = int(input("Enter a number: "))

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print("Reversed number is:", rev)
```

Write a program to check whether a given number is a palindrome or not using a while loop

```python
#WAP to check whether a given number is a palindrome or not using a while loop
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```
