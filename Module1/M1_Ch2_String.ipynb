##**String** **Method**

Write a program to apply all the string method

```python
#string methods
s1 = "umang"
print(s1.upper())             #covert all the character in capital letter
print(s1.lower())             #covert all the character in small letter
print(s1.title())             #convert first letter of string in capital letter
print(s1.capitalize())        #convert first letter of a line in capital letters
print(s1.swapcase())          #convert first letter of a line in capital letters
print(s1.find("h"))           #find the character in the string
print(s1.index("a"))          #find the indexing of a character in string
print(s1.count("h"))          #find the count of a single string in whole string
print(s1.startswith("i"))     #check the string starts with the letter
print(s1.endswith("ka"))      #check the string ends with the letter
```
