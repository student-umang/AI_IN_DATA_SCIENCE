Write a Python program to open a text file umangfile.txt in read mode, read the first three lines, remove extra spaces/newline characters using the strip() method

```python
#remove extra space using (strip)
f = open("Umangfile.txt","r")
Line1=f.readline().strip()
print(Line1)
Line2 = f.readline().strip()
print(Line2)
Line3 = f.readline().strip()
print(Line3)
```

Write a Python program to read the first three lines of a text file and print them without adding extra blank lines between the lines. Use the end="" parameter.

```python
#method  2 remove extra space using (end=" ")
f = open("Umangfile.txt","r")
Line1=f.readline()
print(Line1,end=" ")
Line2 = f.readline()
print(Line2,end=" ")
Line3 = f.readline()
print(Line3,end=" ")
```

##Exception handling


Write a Python program to demonstrate exception handling using try, except, and finally. Handle the ZeroDivisionError that occurs when dividing 10 by 0, print "error" when the exception occurs, and always print "execute" using the finally block.

```python
try:
  print(10//0)
except ZeroDivisionError:
  print("error")
finally:
  print("execute")
```
