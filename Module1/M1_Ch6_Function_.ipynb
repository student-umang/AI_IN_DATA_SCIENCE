Write a program to find the factorial of a number using function

```python
#factorial of number using function
n=int(input("enter number"))
def factorial(n):
  fact=1
  for i in range(1,n+1):
    fact=fact*i
  return fact


factorial(n)
```

Write a program to find the factorial of a number using global and local variable

```python
#local and global varial
n=int(input("enter number")) #global variable
def factorial(n):
  fact=1                    #local variable
  for i in range(1,n+1):
    fact=fact*i
  return fact


factorial(n)
```
