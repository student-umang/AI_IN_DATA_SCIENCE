 Write a program to separate the even and odd numbers from a given list and store them in two separate lists

```python
#WAP to separate the even and odd numbers from a given list and store them in two separate lists
numbers = [10, 15, 20, 25, 30, 35]

even = []
odd = []

for i in numbers:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even Numbers:", even)

print("Odd Numbers:", odd)
```

**Shallow copy**

Write a Python program to demonstrate the concept of shallow copy. Create a nested list, make a shallow copy of it using the copy() method, modify an element in the copied list, and display both the original and copied lists to show the effect of shallow copying.

```python
#shallow copy
import copy

# Original list
list1 = [[1, 2], [3, 4]]

# Creating shallow copy
list2 = copy.copy(list1)

print("Original list:", list1)
print("Copied list:", list2)

# Changing nested list
list2[0][0] = 100

print("\nAfter changing copied list:")
print("Original list:", list1)
print("Copied list:", list2)
```

**Tuple**

Write a program to esxtract and print all even numbers from a given tuple

```python
#WAP to print print even tuple
numbers = (1, 2, 3, 4, 5, 6, 7, 8)

even_tuple = ()

for i in numbers:
    if i % 2 == 0:
        even_tuple += (i,)

print("Even Tuple:", even_tuple)
```
