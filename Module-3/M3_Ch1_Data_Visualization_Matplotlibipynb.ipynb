```python
# Import matplotlib for creating graphs
import matplotlib.pyplot as plt

#marks and study of hours
marks = [98,60,52,72,80]
std_hours = [10, 6, 5, 7, 8]

#Add a plot
plt.plot(marks, std_hours)

for i in range(len(marks)):
  plt.text(marks[i],std_hours[i], str([[marks[i], std_hours[i]]]))

#Add a title
plt.title("marks of studying")

#Add label to x-axis
plt.xlabel("marks")

#Add label to y-axis
plt.ylabel("std_hours")

#display result
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample data for Month and Profit
data = {"Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul"],
        "Profit": [100, 120, 150, 130, 180, 160, 200]}
df = pd.DataFrame(data)

plt.plot(df["Month"], df["Profit"], marker="o")


#add title
plt.title("Month vs Profit")

#add y label
plt.xlabel("Month")

#add y label
plt.ylabel("Profit")


plt.xticks(rotation=45)

#display result
plt.show()
```

```python
# Import matplotlib for creating graphs
import matplotlib.pyplot as plt

# Import numpy for mathematical calculations
import numpy as np

age = [20, 19, 8, 19, 16, 20, 21, 23, 28, 19, 23]


plt.hist(age, bins=3, edgecolor='r', color='yellow', alpha = 0.5)

# Add a title to the histogram
plt.title("Age comparison", fontsize=22, color="green")

# Add a label to the x-axis
plt.xlabel("age", fontsize=12, color="red")

# Add horizontal grid lines
plt.grid(axis='y')

# Add a label to the y-axis
plt.ylabel("", fontsize=12, color="red")

# Calculate the median of the age values
median_age = np.median(age)

# Draw a vertical dashed line at the median value
# linestyle='--' makes the line dashed
# linewidth=3.5 controls the thickness of the line
plt.axvline(median_age,color='blue',linestyle='--',linewidth=3.5,label=f'Median = {median_age}')

# Display the legend
plt.legend()

# Display the histogram
plt.show()
```

**3.3 Bar Chart**

```python
#Import matplotlib for creating graphs
import matplotlib.pyplot as plt


#subject and their score
sub = ["python", 'c', 'java', 'DSA']
score = [40, 50, 60, 70]

#Used to display bar chart
plt.bar(sub, score, color=["red", "green", "yellow", "pink"])

#Add a title in a bar chart
plt.title("Subject vs Score", fontsize = 18)

#Add horizontal grid lines
plt.grid(axis='y')

#Add a label to x-axis in bar chart
plt.xlabel("Subject", color = 'g')

#Add a label to y-xis in bar chart
plt.ylabel("Score", color = 'g')

#Display the bar chart
plt.show()
```

**3.4 Pie Chart**

```python
#subject and their score
sub = ['technical', 'engkish', 'gk', 'apptitude']
score = [50, 20, 20, 10]

#Used to create a pie chart
plt.pie(score, labels = sub, autopct = "%1.2f%%")

#Add a title to a pie chart
plt.title("Subject vs Score")

#Display the pie chart
plt.show()
```

**3.5 Scatter Plot**

```python
# Data from 1st scatter plot
x1 = [2, 4, 6, 8, 100]
y1 = [11, 13, 15, 17, 19]


# Data from 2st scatter plot
x2 = [12, 14, 16, 18, 90]
y2 = [1, 3, 5, 7, 9]


#define color and size of data points
rang = ['pink', 'purple', 'orange', 'red', 'green']
size = [100, 200, 300, 400, 500]


#create 1st and 2nd scatter plot
plt.scatter(x1,y1, color=rang, s = size, alpha = 0.5)
plt.scatter(x2,y2, color=rang, s = size, alpha = 0.8)


#display the legend
plt.legend()

#add color bar to represent intensity
plt.colorbar(label = "intensity")

#display the final plot
plt.show()
```
