```python
# Capitalizes the first letter of each column name
import pandas as pd

california_housing_test_df = pd.read_csv('/content/sample_data/california_housing_test.csv')

california_housing_test_df.columns = california_housing_test_df.columns.str.capitalize()
print(california_housing_test_df.columns)
```

```python
#Remove leading and trailing spaces fromm all text columns
for column in california_housing_test_df.select_dtypes(
    include="object"
).columns:

    california_housing_test_df[column] = (
        california_housing_test_df[column]
        .str.strip()
    )
    print(california_housing_test_df.head())
    print(california_housing_test_df.select_dtypes(include="object").columns)
```

```python
# Convert each column into a numeric data type
# errors='coerce' converts invalid values into NaN

for column in california_housing_test_df.columns:
    california_housing_test_df[column] = pd.to_numeric(
        california_housing_test_df[column],
        errors='coerce'
    )

# Display the updated data types
print(california_housing_test_df.dtypes)
```

```python
# Fill missing values in numeric columns with their median
# Median is useful because it is less affected by extreme values

for column in california_housing_test_df.select_dtypes(
    include='number'
).columns:

    california_housing_test_df[column] = (
        california_housing_test_df[column]
        .fillna(california_housing_test_df[column].median())
    )

# Check missing values again
print(california_housing_test_df.isnull().sum())
```

```python
# Fill missing values in numeric columns with their median
# Median is useful because it is less affected by extreme values

for column in california_housing_test_df.select_dtypes(
    include='number'
).columns:

    california_housing_test_df[column] = (
        california_housing_test_df[column]
        .fillna(california_housing_test_df[column].median())
    )

# Check missing values again
print(california_housing_test_df.isnull().sum())
```

```python
# Count zero values in each column
# This helps identify values that may cause problems
# when creating ratios or performing mathematical transformations

print((california_housing_test_df == 0).sum())
```

```python
# Count negative values in each column
# This helps identify potentially invalid values

print((california_housing_test_df < 0).sum())
```

```python
# Select only numerical columns for outlier detection

numeric_columns = california_housing_test_df.select_dtypes(
    include='number'
).columns

# Calculate the first quartile (Q1)
Q1 = california_housing_test_df[numeric_columns].quantile(0.25)

# Calculate the third quartile (Q3)
Q3 = california_housing_test_df[numeric_columns].quantile(0.75)

# Calculate Interquartile Range (IQR)
IQR = Q3 - Q1

# Calculate lower and upper limits for detecting outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:")
print(lower_bound)

print("\nUpper Bound:")
print(upper_bound)
```

```python
# Count the number of outliers in each numerical column

outliers = (
    (california_housing_test_df[numeric_columns] < lower_bound) |
    (california_housing_test_df[numeric_columns] > upper_bound)
)

print("Number of outliers in each column:")
print(outliers.sum())
```

```python
import matplotlib.pyplot as plt

# Create a boxplot to visually identify outliers

california_housing_test_df[numeric_columns].plot(
    kind='box',
    figsize=(15, 7)
)

# Rotate column names for better readability
plt.xticks(rotation=45)

# Add title to the plot
plt.title("Boxplot for Outlier Detection")

# Display the plot
plt.show()
```

```python
# Create a new feature showing the average number of rooms
# available per household

california_housing_test_df['rooms_per_household'] = (
    california_housing_test_df['total_rooms'] /
    california_housing_test_df['households']
)

# Display the new column
print(california_housing_test_df[
    ['total_rooms', 'households', 'rooms_per_household']
].head())
```
