##** MERGE**

1) **Merge**

```python
import pandas as pd
df1= pd.DataFrame({"A":[2,3,4,5],"B":["a","b","c","d"]})
df2= pd.DataFrame({"A":[2,4,6,8],"B":["a","c","e","f"]})


# By default, pandas performs an inner merge on common columns (here, 'A' and 'B').
pd.merge(df1,df2)
```

```python
import pandas as pd
df1= pd.DataFrame({"A":[2,3,4,5],"B":["a","b","c","d"]})
df2= pd.DataFrame({"A":[2,4,6,8],"B":["a","c","e","f"]})
pd.merge(df1,df2)
df3=pd.merge(df1,df2)
print(df3)        # Display the merged DataFrame
```

```python
import pandas as pd
df1= pd.DataFrame({"A":[2,3,4,5],"B":["a","b","c","d"]})# Merges df1 and df2 on the common column 'A'.
 # It performs an inner merge by default, only keeping rows where 'A' values match in both
 # Columns 'B' and 'C' are included as they are unique to each DataFrame.
df2= pd.DataFrame({"A":[2,4,6,8],"C":["a","c","e","f"]})
pd.merge(df1,df2)
```

```python
import pandas as pd
df1= pd.DataFrame({"A":[2,3,4,5],"B":["a","b","c","d"]})
df2= pd.DataFrame({"A":[2,4,6,8],"B":["a","c","e","f"]})
pd.merge(df1,df2)
pd.merge(df1,df2, on ="A",how ="inner")

```

##**JOIN**

```python
import pandas as pd
df1= pd.DataFrame({"A":[2,3,4,5],"B":["a","b","c","d"]})
df2= pd.DataFrame({"D":[2,4,6,8],"C":["a","c","e","f"]})
df1.join(df2)
```

```python
import pandas as pd
df1= pd.DataFrame({"A":[2,3,4,5],"B":["a","b","c","d"]},index=["a","b","c","d"])
df2= pd.DataFrame({"C":[2,4,6,8],"D":["a","c","e","f"]})
df1.join(df2)
```
