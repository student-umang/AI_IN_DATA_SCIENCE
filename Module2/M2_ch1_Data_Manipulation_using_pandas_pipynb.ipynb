Create a Pandas DataFrame with missing values and use dropna(subset=["marks"]) to remove only those rows where the marks column contains a missing value. Display the resulting DataFrame.

```python
import numpy as np
import pandas as pd

std = {
    "name": ["kanishka", "umang", "sadaf", "kunal", "naina"],
    "roll_no.": [10, 11, 12, 15, np.nan],
    "marks": [20, np.nan, 30, 60, 50],
    "subject": ["hindi", "english", np.nan, "maths", "physics"],
    "city": ["delhi", np.nan, "bijnor", "muzaffarnaar", "chandigarh"],
}

df = pd.DataFrame(std)

# Drop rows where the 'marks' column has missing values
df = df.dropna(subset=["marks"])

print(df)
```
